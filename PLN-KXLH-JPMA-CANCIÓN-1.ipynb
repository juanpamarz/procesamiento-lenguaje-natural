String es nátivo de python y sirve como una librería que contiene los signos de puntuación para poder eliminarlos fácilmente. <p>
Counter es importado de collection para poder contar las palabras más fácilmente sin necesidad de hacer mi propio contador. <p>
Stopwords importa las palabras que actuan como relleno, como podrían ser conectores, artículos y preposiciones. <p>
Word_tokenize Se encarga de tokenizar cada una de las palabras del texto.

```python
import string
from collections import Counter
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('stopwords')
nltk.download("punkt_tab")
```

Importamos la canción de nuestros archivos y le damos el encoding 'utf-8' para que reconozca tildes y la ñ. 

```python
cancion_1 = r"C://Users//jpma2//Documents//UNIVERSIDAD//mineria de textos//prácticas//procesamiento de lenguaje natural//cancion 1.txt"
with open(cancion_1, 'r', encoding='utf-8') as file:
    letra = file.read()
print (letra)
```

Tokenizamos y hacemos minusculas todas las palabras, para que no haga distinción entre 'Beber' y 'beber' por dar un ejemplo. 

```python
tokens = word_tokenize(letra.lower())
```

Recorremos la canción ya tokenizada y hacemos que no tome en cuenta todos los signos de puntuación tradicionales en el inglés y como en español también usamos los signos de apertura los añadimos a la instrucción. 

```python
tokens_sin_puntuacion = [
    word for word in tokens if word not in string.punctuation and word not in "¿¡"
]
```

Imprimimos como se ven las palabras ya tokenizadas sin signos de puntuación

```python
print (tokens_sin_puntuacion)
```

Recorremos el arreglo para eliminar los conectores desde la lista de palabras de la librería que ya vienen configuradas para nuestro idioma. 

```python
stop_words = set(stopwords.words("spanish"))
palabras_filtradas = [
    word for word in tokens_sin_puntuacion if word not in stop_words
]
print (palabras_filtradas)
```
