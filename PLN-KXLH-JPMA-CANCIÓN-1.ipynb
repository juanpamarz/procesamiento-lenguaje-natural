String es nátivo de python y sirve como una librería que contiene los signos de puntuación para poder eliminarlos fácilmente. <p>
Counter es importado de collection para poder contar las palabras más fácilmente sin necesidad de hacer mi propio contador. <p>
Stopwords importa las palabras que actuan como relleno, como podrían ser conectores, artículos y preposiciones. <p>
Word_tokenize Se encarga de tokenizar cada una de las palabras del texto.<p>
Matplotlib.pyplot sirve para construir las gráficas. 

```python
import string
from collections import Counter
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import matplotlib.pyplot as plt
#Las descargas de nltk se hacen una sola vez, si ya se han descargado no es necesario volver a hacerlo
nltk.download('punkt')
nltk.download('stopwords')
nltk.download("punkt_tab")
```

Importamos la canción de nuestros archivos y le damos el encoding 'utf-8' para que reconozca tildes y la ñ. 

```python
cancion_1 = r"C://Users//jpma2//Documents//UNIVERSIDAD//mineria de textos//prácticas//procesamiento de lenguaje natural//cancion 1.txt"
with open(cancion_1, 'r', encoding='utf-8') as file:
    letra = file.read()
print (letra)
```

Tokenizamos y hacemos minusculas todas las palabras, para que no haga distinción entre 'Beber' y 'beber' por dar un ejemplo. 

```python
tokens = word_tokenize(letra.lower())
```

Recorremos la canción ya tokenizada y hacemos que solo tome en cuenta palabras, nada de signos. 

```python
tokens_sin_puntuacion = [
    word
    for word in tokens
    if word.isalpha()
]
```

Imprimimos como se ven las palabras ya tokenizadas sin signos de puntuación

```python
print (tokens_sin_puntuacion)
```

Recorremos el arreglo para eliminar los conectores desde la lista de palabras de la librería que ya vienen configuradas para nuestro idioma. 

```python
stop_words = set(stopwords.words("spanish"))
palabras_filtradas = [
    word for word in tokens_sin_puntuacion if word not in stop_words
]
print (palabras_filtradas)
```

Hacemos un contador que nos diga cuántas veces se repitio cada palabra. 

```python
Contador = Counter(palabras_filtradas)
for palabra, frecuencia in Contador.most_common():
  print(f"- {palabra}: {frecuencia}")
```

Graficamos las 10 palabras más comunes en el texto. 

```python
palabras = [palabra for palabra, frecuencia in Contador.most_common(10)]
frecuencias = [frecuencia for palabra, frecuencia in Contador.most_common(10)]

plt.bar(palabras, frecuencias)
plt.xlabel("Palabras")
plt.ylabel("Frecuencia")
plt.title("Palabras más frecuentes en la canción")
plt.xticks(rotation=80)
plt.show()
```

Creamos un diccionario para hacer categorías de las palabras

```python
categorias = {
    "Ave": [
        "gorrión", "jaula", "alpiste", "volar",
        "cielo", "alas", "canto"
    ],

    "Alimentación": [
        "beber", "agua", "fruto", "dabas",
    ],

    "Relación": [ 
        "ofreció", "ofreces", "dale", "dejaste", "olvidado",
        "confundido", "cansé", "darte", "poco", "luz",
        "pude", "completo", "anhelado","vuelvas", "buscarme", "vete", "deja",
        "caiga", "quedó", "preso", "corté",
        "déjame", "encender", "quede","ganaría", "tomaría"
    ],
    "Amor": [
        "amaba", "amarte", "quiero"
    ],

    "Tiempo": [
        "día", "vez", "ayer", "primera", "sido", "siempre"
    ],

    "S/C": [
        "manos", "mano", "aquí", 
        "solo", "abierta", "tan", "embargo", 
    ]

}
```

```python
frecuencia_categorias = {}

for categoria, words in categorias.items():
    frecuencia = sum(Contador[word] for word in words)
    frecuencia_categorias[categoria] = frecuencia


# Mostrar resultados
print("Frecuencia por categoría:\n")

for categoria, frecuencia in frecuencia_categorias.items():
    print(f"- {categoria}: {frecuencia}")


# Crear gráfica
plt.figure(figsize=(10, 6))

plt.bar(
    frecuencia_categorias.keys(),
    frecuencia_categorias.values()
)

plt.xlabel("Categorías")
plt.ylabel("Frecuencia")
plt.title("Frecuencia de categorías en la canción")

plt.xticks(rotation=90, ha="right")
plt.show()
```
